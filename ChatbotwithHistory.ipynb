```python
import os
from dotenv import load_dotenv
load_dotenv() # loading all the environment variables

groq_api_key=os.getenv("GROQ_API_KEY")
groq_api_key
```

```python
from langchain_groq import ChatGroq
model=ChatGroq(model="openai/gpt-oss-20b",groq_api_key=groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage
model.invoke([HumanMessage(content="Hi, My name is Nikitaa  and I am a AI Engineering Student")])
```

```python
from langchain_core.messages import AIMessage
model.invoke(
    [
        HumanMessage(content="Hi, My name is Nikita and I am a student at MSIT"),
        AIMessage(content="Hello Nikita! Its nice to interact to you. What course are you pursuing?"),
        HumanMessage(content="Hey, Whats my name and what do i do?")
    ]
)
```

```python
from langchain_core.chat_history import (BaseChatMessageHistory, InMemoryChatMessageHistory)
from langchain_core.runnables.history import RunnableWithMessageHistory

store={} # how will you know different chat session is a for different chat history
def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=InMemoryChatMessageHistory()
    return store[session_id]      
```
